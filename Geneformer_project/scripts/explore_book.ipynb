```python
def find_human_entrez_id(homologene_genes_list):
    for species_id, gene_id in homologene_genes_list:
        if species_id == 9606:
            return gene_id
    return None 
```

```python
mouse_to_human_dicts = {}
for r in human_orthologs:
    word = r.get('homologene', {}) if isinstance(r.get('homologene'), dict) else None
    if word is not None:
        mouse_to_human_dicts[r["query"]] = find_human_entrez_id(word["genes"])
```
